## 1.2 Drop the Unnamed Columns

```python
df_tb = pd.read_csv('API_SH.TBS.INCD_DS2_en_csv_v2_35263.csv', skiprows=4)
# Drop columns that start with 'Unnamed:'
df_tb = df_tb.loc[:, ~df_tb.columns.str.contains('^Unnamed')]

print("DataFrame after re-loading and re-dropping 'Unnamed' columns:")
print(df_tb.head())
```

## 1.3 Double-check all columns' datatype

```python
print("DataFrame Info (including dtypes):")
df_tb.info()
```

## 1.4 Display all numeric columns with the same format of two decimal points

```python
# Set display option to format float values to two decimal points
pd.options.display.float_format = '{:.2f}'.format

print("DataFrame with numeric columns formatted to two decimal points:")
display(df_tb.head())
```

## 1.5 Drop *columns* which contain *only* NaN values

```python
df_tb = df_tb.dropna(axis=1, how='all')

print("DataFrame after dropping columns with only NaN values:")
print(df_tb.head())
print(f"New DataFrame shape: {df_tb.shape}")
```

# Task 2 - Indexing and Aggregating

## 2.1 Find the TB incidence of the US in 2019. Show only the relevant number.

```python
us_incidence_2019 = df_tb[df_tb['Country Name'] == 'United States']['2019'].iloc[0]
print(f"TB incidence of the US in 2019: {us_incidence_2019:.2f}")
```

## 2.2 What is the average incidence of the US in the 2010s (from 2010 to 2019)?

```python
us_incidence_2010s = df_tb[df_tb['Country Name'] == 'United States'][['2010', '2011', '2012', '2013', '2014', '2015', '2016', '2017', '2018', '2019']].mean(axis=1).iloc[0]
print(f"Average TB incidence of the US in the 2010s: {us_incidence_2010s:.2f}")
```

## 2.3 In which year did the US experience the highest incidence? Check `idxmax`.

```python
us_data = df_tb[df_tb['Country Name'] == 'United States']
year_cols = [col for col in df_tb.columns if col.isdigit()]
highest_incidence_year_us = us_data[year_cols].idxmax(axis=1).iloc[0]
highest_incidence_value_us = us_data[year_cols].max(axis=1).iloc[0]

print(f"The US experienced the highest TB incidence in {highest_incidence_year_us} with a value of {highest_incidence_value_us:.2f}")
```

## 2.4 Which country had the highest incidence in 2024? What is that value?

```python
highest_incidence_country_2024 = df_tb.loc[df_tb['2024'].idxmax()]
highest_country_name = highest_incidence_country_2024['Country Name']
highest_country_incidence = highest_incidence_country_2024['2024']

print(f"The country with the highest TB incidence in 2024 is {highest_country_name} with a value of {highest_country_incidence:.2f}")
```

## 2.5 Which country experienced the sharpest decrease in incidence from 2014 to 2015? What is that value?

```python
df_tb['Decrease_2014_2015'] = df_tb['2014'] - df_tb['2015']
sharpest_decrease_country = df_tb.loc[df_tb['Decrease_2014_2015'].idxmax()]
country_name_sharpest_decrease = sharpest_decrease_country['Country Name']
decrease_value = sharpest_decrease_country['Decrease_2014_2015']

print(f"The country with the sharpest decrease in incidence from 2014 to 2015 is {country_name_sharpest_decrease} with a decrease of {decrease_value:.2f}")
```

# Task 3 - Joining / Merging

## 3.1 Load the meta file

```python
df_meta = pd.read_csv(
    'Metadata_Country_API_SH.TBS.INCD_DS2_en_csv_v2_35263.csv',
    skiprows=4,
    header=None
)

# Assign the correct column names
df_meta.columns = [
    'Country Code',
    'Region',
    'IncomeGroup',
    'Special Notes',
    'Country Name',
    'Unnamed'
]

# Drop the unnecessary column
df_meta.drop(columns=['Unnamed'], inplace=True)

print(df_meta.head())
print(df_meta.columns.tolist())
```

```python
# Drop the Unnamed Column
```

# 3.2 Merge with the TB DataFrame by CountryCode. Keep only the Country Code, Region, IncomeGroup, and the yearly Incidence columns.

```python
year_cols = [col for col in df_tb.columns if col.isdigit()]

columns_to_keep = ['Country Code', 'Region', 'IncomeGroup'] + year_cols

df_merged = pd.merge(
    df_tb,
    df_meta[['Country Code', 'Region', 'IncomeGroup']],
    on='Country Code',
    how='inner'
)

df_merged = df_merged[columns_to_keep]

print("Merged DataFrame (df_merged) head:")
print(df_merged.head())

print(f"Merged DataFrame shape: {df_merged.shape}")
```

## 3.3 What is the average incidence by Region in 2024? Use `.groupby()`

```python
avg_incidence_by_region_2024 = df_merged.groupby('Region')['2024'].mean()

print("Average TB incidence by Region in 2024:")
print(avg_incidence_by_region_2024.to_string())
```

## 3.4 What is the average incidence by IncomeGroup in 2024?

```python
avg_incidence_by_incomegroup_2024 = df_merged.groupby('IncomeGroup')['2024'].mean()

print("Average TB incidence by Income Group in 2024:")
print(avg_incidence_by_incomegroup_2024.to_string())
```

## 3.5 Which high-income country has the highest incidence in 2024?

```python
high_income_countries = df_merged[
    df_merged['IncomeGroup'] == 'High income'
]

highest_incidence_high_income_2024 = high_income_countries.loc[
    high_income_countries['2024'].idxmax()
]

country_code = highest_incidence_high_income_2024['Country Code']

country_name = df_tb.loc[
    df_tb['Country Code'] == country_code,
    'Country Name'
].iloc[0]

incidence_value = highest_incidence_high_income_2024['2024']

print(
    f"The high-income country with the highest TB incidence in 2024 "
    f"is {country_name} with an incidence of {incidence_value:.2f}"
)
```
